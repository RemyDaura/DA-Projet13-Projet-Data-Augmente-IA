```python
#Création d'une colonne "stock_status_2"
#La valeur de cette deuxième colonne sera fonction de la valeur dans la colonne "stock_quantity"
#Si la valeur de la colonne "stock_quantity" est nulle, renseigner "outofstock" sinon mettre "instock"
# 2. Création de la colonne "stock_status_2"
# Si 'stock_quantity' est <= 0, alors 'outofstock', sinon 'instock'
df_erp['stock_status_2'] = np.where(
    df_erp['stock_quantity'] <= 0,
    'outofstock',
    'instock'
)

# 3. Afficher les premières lignes pour vérifier le résultat
print("--- Vérification de la nouvelle colonne 'stock_status_2' ---")
print(df_erp[['stock_quantity', 'stock_status', 'stock_status_2']].head(10))

# 4. Afficher le nombre de différences pour l'analyse
differences = (df_erp['stock_status'] != df_erp['stock_status_2']).sum()
print(f"\nNombre de lignes où 'stock_status' et 'stock_status_2' diffèrent : {differences}")
```

```python
#Vérifions que les 2 colonnes sont identiques:
#Les 2 colonnes sont strictement identiques si les valeurs de chaque ligne sont strictement identiques 2 à 2
#La comparaison de 2 colonnes peut se réaliser simplement avec l'instruction ci-dessous:
df_erp["stock_status"] == df_erp["stock_status_2"]

#Le résultat est l'affichage de True ou False pour chacune des lignes du dataset
#C'est un bon début, mais difficile à exploiter
```

```python
#Mais il est possible de synthétiser ce résultat en effectuant la somme de cette colonne:
#True vaut 1 et False 0
#Nous devrions obtenir la somme de 824 qui correspond au nombre de lignes dans ce dataset
comparaison = (df_erp["stock_status"] == df_erp["stock_status_2"])
somme_comparaison = comparaison.sum()
print(somme_comparaison)
```

```python
#Si les colonnes ne sont absolument pas identiques ligne à ligne alors identifier la ligne en écart
##Dans ce cas je vous donne ce lien pour apprendre à réaliser des filtres dans Pandas:
##https://bitbucket.org/hrojas/learn-pandas/src/master/
##Lesson 3
df_erp[df_erp["stock_status"] != df_erp["stock_status_2"]]
```

```python
#Corriger la ou les données incohérentes
df_erp['stock_status'] = df_erp['stock_status_2']
df_erp = df_erp.drop(columns=['stock_status_2'])
#Vérification en utilisant le même code que plus haut pour afficher les problèmes
verification_finale = df_erp[df_erp['stock_status'] != np.where(df_erp['stock_quantity'] <= 0, 'outofstock', 'instock')]

print("--- Vérification finale des lignes en écart après correction ---")
print(verification_finale[['product_id', 'stock_quantity', 'stock_status']])
```

```python
#Fusion des fichiers df_erp et df_liaison
# 1. Normalisation des types de données pour la clé de jointure 'product_id'
# C'est essentiel pour éviter les erreurs de type (int64 vs object)
df_erp['product_id'] = df_erp['product_id'].astype(str)
df_liaison['product_id'] = df_liaison['product_id'].astype(str)

# Fusion des fichiers df_erp et df_liaison
# Jointure GAUCHE : On garde tous les produits de l'ERP (df_erp)
df_erp_liaison = pd.merge(
    df_erp,
    df_liaison,
    on='product_id',
    how='left'
)
```

```python
# Filtrer les lignes où 'id_web' est NaN après la jointure (ce sont les articles de l'ERP sans ID Web)
lignes_non_matchantes = df_erp_liaison[df_erp_liaison['id_web'].isnull()]

print(f"Nombre de lignes de l'ERP sans correspondance Web : {len(lignes_non_matchantes)}")
print("\n--- Détail des produits ERP sans ID Web (5 premières lignes) ---")
print(lignes_non_matchantes[['product_id', 'price', 'purchase_price', 'stock_quantity']].head())
```

```python
# Suppression des produits ERP sans lien Web
df_final_clean_clean = df_erp_liaison.dropna(subset=['id_web']).copy()

print(f"\nNombre de lignes conservées pour la suite (après suppression) : {len(df_final_clean_clean)}")
```

```python
# S'assurer que les clés sont de type string
df_erp_liaison['id_web'] = df_erp_liaison['id_web'].astype(str)
df_web['sku'] = df_web['sku'].astype(str)

# Fusionner les datasets df_erp_liaison et df_web
# Jointure GAUCHE : On garde tous les produits ayant un prix/stock
df_final = pd.merge(
    df_erp_liaison,
    df_web,
    left_on='id_web',
    right_on='sku',
    how='left'
)
```

```python
# Suppression des doublons basée sur 'sku', en gardant la première occurrence
df_web = df_web.drop_duplicates(subset=['sku'], keep='first')

# Vérification du nombre de doublons restants
nombre_doublons_sku = df_web['sku'].duplicated().sum()

print(f"Nombre de lignes en double (doublons) dans df_web après nettoyage : {nombre_doublons_sku}")
```

```python
# Identifier les lignes où les données Web sont manquantes après la jointure
# On vérifie si la colonne 'post_title' (qui vient de df_web) est nulle.
lignes_web_manquantes = df_final[df_final['post_title'].isnull()]

print(f"Nombre de produits ERP/Liaison sans fiche Web correspondante : {len(lignes_web_manquantes)}")
print("\n--- Détail des produits sans titre Web (ou fiche Web) ---")
print(lignes_web_manquantes[['product_id', 'id_web', 'post_title', 'total_sales']].head())
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# 1. Chargement et Nettoyage
df_erp = pd.read_csv('erp.csv', sep=';', decimal=',', encoding='latin-1')
df_web = pd.read_csv('web.csv', sep=';', decimal=',', encoding='latin-1')
df_liaison = pd.read_csv('liaison.csv', sep=';', encoding='latin-1')

# Préparation Web
df_web = df_web[df_web['post_type'] == 'product']
df_web = df_web.rename(columns={'sku': 'id_web'})

# Typage des clés
df_web['id_web'] = df_web['id_web'].astype(str)
df_liaison['id_web'] = df_liaison['id_web'].astype(str)

# Fusions
df_temp = pd.merge(df_liaison, df_web, on='id_web', how='inner')
df_final = pd.merge(df_temp, df_erp, on='product_id', how='inner')

# 2. Calculs des Nouvelles Variables
# Conversion numérique de base
for col in ['price', 'purchase_price', 'total_sales', 'stock_quantity']:
    df_final[col] = pd.to_numeric(df_final[col], errors='coerce')

# Calcul du Prix HT (TVA 20%)
df_final['price_ht'] = df_final['price'] / 1.2

# Calcul du Taux de Marge
# Formule : (Prix de vente - Coût d'achat) / Coût d'achat
df_final['taux_marge'] = (df_final['price'] - df_final['purchase_price']) / df_final['purchase_price']

# 3. Sélection des 6 colonnes pour la heatmap
cols_corr = ['stock_quantity', 'total_sales', 'price', 'taux_marge', 'price_ht', 'purchase_price']

# Calcul de la matrice
corr_matrix = df_final[cols_corr].corr()

# 4. Création du Masque (Triangle supérieur)
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

# 5. Affichage Graphique
plt.figure(figsize=(10, 8)) # Taille augmentée pour la lisibilité
sns.heatmap(
    corr_matrix, 
    mask=mask, 
    annot=True, 
    fmt=".2f", 
    cmap='coolwarm', # Palette divergente (Bleu <-> Rouge)
    vmin=-1, 
    vmax=1,
    center=0,
    square=True,
    linewidths=.5,
    cbar_kws={"shrink": .5}
)

plt.title('Matrice de Corrélation Complète (6 Variables)')
plt.show()
```

```python
#Que peut-on conclure des corrélations ?
#Corrélation Prix et Ventes : La valeur est proche de 0, cela signifie que le prix n'a pas d'influence linéaire sur la quantité vendue. 
#Cela peut indiquer que le catalogue est dominé par des vins de collection ou de niche, où le prix est justifié par la rareté 
#et la demande est inélastique.
#Corrélation Stock et Ventes : Idéalement, cette corrélation devrait être faible. Une corrélation élevée signifierait que vous stockez beaucoup 
#plus là où vous vendez beaucoup (risque d'excès de stock) ou que vous manquez de stock là où la demande est forte.
#Corrélation Stock et Prix : Si la valeur est positive, cela signifie que vous immobilisez votre capital sur les produits les plus chers. 
#Étant donné que ces produits ont souvent un faible volume de vente, cela confirme le risque identifié dans le FLOP 20 : 
#le capital dormant est concentré sur les articles de luxe, ce qui est un usage inefficace du capital si ces produits ne se vendent pas.
```

### 1) Chargement des donnees brutes (sans nettoyage prealable)

```python
import pandera.pandas as pa
from pandera.pandas import Column, Check, DataFrameSchema

df_erp_brut = pd.read_csv("erp.csv", sep=";", decimal=",", encoding="latin-1")
df_web_brut = pd.read_csv("web.csv", sep=";", decimal=",", encoding="latin-1")
df_liaison_brut = pd.read_csv("liaison.csv", sep=";", encoding="latin-1")

print(f"{len(df_erp_brut)} lignes ERP / {len(df_web_brut)} lignes Web / {len(df_liaison_brut)} lignes liaison (brutes)")
```

### 2) Schema de validation - fichier ERP (Pandera)

Trois regles suffisent a couvrir les anomalies "Prix Errones" et une partie
de "Incoherence Stock" identifiees manuellement plus haut dans ce notebook.

```python
schema_erp = DataFrameSchema({
    "product_id": Column(int, Check.greater_than(0), unique=True, nullable=False),
    "onsale_web": Column(int, Check.isin([0, 1])),
    "price": Column(float, Check.greater_than(0), nullable=False),
    "stock_quantity": Column(int, Check.greater_than_or_equal_to(0)),
    "stock_status": Column(str, Check.isin(["instock", "outofstock"])),
    "purchase_price": Column(float, Check.greater_than(0), nullable=False),
}, strict=False)

try:
    schema_erp.validate(df_erp_brut, lazy=True)
    print("Aucune anomalie de schema.")
except pa.errors.SchemaErrors as err:
    print(f"{len(err.failure_cases)} anomalies detectees automatiquement :\n")
    display(err.failure_cases[["column", "check", "failure_case", "index"]])
```

```python
# Regle croisee : coherence entre stock_status et stock_quantity
# [Version initiale] ne testait qu'un seul sens de la regle (instock + qte<=0).
# [Corrige apres test de la methode 3] l'IA directe a trouve un 2e cas dans
# l'autre sens (outofstock + qte>0, ID 4039) que cette premiere version ratait.
incoherents_instock = df_erp_brut[(df_erp_brut["stock_status"] == "instock") & (df_erp_brut["stock_quantity"] <= 0)]
incoherents_outofstock = df_erp_brut[(df_erp_brut["stock_status"] == "outofstock") & (df_erp_brut["stock_quantity"] > 0)]
incoherents = pd.concat([incoherents_instock, incoherents_outofstock])

print(f"Incoherences stock_status/stock_quantity (2 sens) : {len(incoherents)} ligne(s)")
display(incoherents[["product_id", "stock_status", "stock_quantity"]])
```

**Resultat :** les 5 anomalies de schema (3 prix negatifs, 2 stocks negatifs)
sont retrouvees avec les memes product_id que ceux identifies manuellement.
L'incoherence stock_status/stock_quantity, elle, n'etait detectee qu'a moitie
dans la premiere version de ce schema (voir encadre methode 3 plus bas).

### 3) Schema de validation - fichier Web (Pandera)

Le fichier Web brut melange des fiches produit et des lignes techniques
("attachment" = images).

```python
n_attachment = (df_web_brut["post_type"] == "attachment").sum()
n_sku_vide = df_web_brut["sku"].isnull().sum()

print(f"Lignes 'attachment' (hors perimetre produit) : {n_attachment}")
print(f"Lignes avec sku vide                          : {n_sku_vide}")

df_web_products = df_web_brut[df_web_brut["post_type"] == "product"].dropna(subset=["sku"]).copy()
df_web_products["sku"] = df_web_products["sku"].astype(str)

schema_web = DataFrameSchema({
    "sku": Column(str, Check.str_matches(r"^\d+$"), unique=True, nullable=False),
    "post_type": Column(str, Check.equal_to("product")),
}, strict=False)

try:
    schema_web.validate(df_web_products, lazy=True)
    print("\nAucune anomalie de schema sur le perimetre produit.")
except pa.errors.SchemaErrors as err:
    print(f"\n{len(err.failure_cases)} anomalies detectees :\n")
    display(err.failure_cases[["column", "check", "failure_case", "index"]])
```

**Resultat :** les 714 lignes "attachment" et les 85 lignes sans sku sont
retrouvees a l'identique. Le schema signale 2 codes non standards
("13127-1" et "bon-cadeau-25-euros").

### 4) Verifications inter-fichiers

Pandera valide un DataFrame a la fois : les regles qui croisent plusieurs
fichiers sont donc ecrites en pandas simple, mais documentees au meme
endroit que le schema.

```python
# a) "Orphelins" - DEUX causes distinctes, a ne pas fusionner en un seul chiffre.
# [Version initiale] ce POC ne testait que "id_web absent de web.csv" -> 111,
# un chiffre qui ne recoupait pas exactement les 91 documentes lors du P6.
# [Corrige apres test methode 3] l'IA directe a distingue, sans indice prealable,
# 2 causes bien differentes qui expliquent exactement l'ecart :
web_ids_presents = set(df_web_products["sku"])

df_liaison_null = df_liaison_brut[df_liaison_brut["id_web"].isnull()]
print(f"a1) product_id ERP avec id_web NUL dans liaison.csv (jamais relie)    : {len(df_liaison_null)}")

df_liaison_nn = df_liaison_brut.dropna(subset=["id_web"]).copy()
df_liaison_nn["id_web"] = df_liaison_nn["id_web"].astype(str)
orphelins = df_liaison_nn[~df_liaison_nn["id_web"].isin(web_ids_presents)]
print(f"a2) id_web renseigne mais introuvable dans web.csv (lien casse)       : {len(orphelins)}")
print(f"    Total (a1+a2) = {len(df_liaison_null) + len(orphelins)}  <-  correspond exactement au 111 trouve initialement,")
print(f"    et le a1 seul (91) correspond exactement au chiffre documente lors du P6.")
```

```python
# b) Coherence onsale_web (ERP) <-> presence reelle dans le catalogue Web
df_liaison_m = df_liaison_brut.copy()
df_liaison_m["product_id"] = df_liaison_m["product_id"].astype(str)
df_liaison_m["id_web"] = df_liaison_m["id_web"].astype(str)
df_erp_m = df_erp_brut.copy()
df_erp_m["product_id"] = df_erp_m["product_id"].astype(str)

df_check = pd.merge(df_erp_m, df_liaison_m, on="product_id", how="left")
df_check["present_sur_web"] = df_check["id_web"].isin(web_ids_presents)

onsale_mais_absent = df_check[(df_check["onsale_web"] == 1) & (~df_check["present_sur_web"])]
print(f"b) Produits 'onsale_web=1' mais absents du Web : {len(onsale_mais_absent)}")
print("   IDs :", onsale_mais_absent["product_id"].tolist())
```

**Resultat :** les IDs 4594, 5070 et 5075 sont retrouves - coherents avec
les exemples cites lors de la soutenance P6. La decomposition 91 / 20 ci-dessus
recoupe maintenant exactement les chiffres historiques, ce qui n'etait pas le
cas avant la correction (voir methode 3 ci-dessous).

### 5) Methode 3 - Donner les donnees brutes a une IA generative

**Protocole :** dans une conversation neuve (sans lien avec celle utilisee
pour construire ce notebook), les 3 fichiers bruts ont ete transmis a une IA
avec pour seule consigne : *"identifie toutes les anomalies, incoherences ou
erreurs de qualite que tu peux detecter, sans indication prealable sur ce
qu'il faut chercher."* Aucune des regles ci-dessus ne lui a ete communiquee.

**Ce que cette methode a retrouve, a l'identique des methodes 1 et 2 :**
prix negatifs, stocks negatifs, doublons "attachment", ID non standards,
incoherence onsale_web, ventes a perte (7 cas, memes IDs).

**Ce que cette methode a trouve EN PLUS, et que les schemas Pandera/Great
Expectations initiaux ne testaient pas :**
- Le 2e sens de l'incoherence stock_status/quantite (ID 4039) - **corrige
  en section 2 ci-dessus**
- La decomposition exacte 91 (id_web nul) + 20 (lien casse) au lieu d'un
  chiffre agrege de 111 - **corrige en section 4 ci-dessus**
- Un risque de **collision d'identifiants** : les codes non standards
  "13127-1" et "14680-1" contiennent un numero de SKU (13127, 14680) deja
  utilise par un AUTRE produit (4053 et 4757) - jamais teste dans les schemas
- **6 fiches en doublon** (3 titres identiques, sku differents) - categorie
  d'anomalie entierement absente des methodes 1 et 2, et du P6 original
- **2 compteurs de ventes negatifs** (total_sales < 0) - jamais teste

**Ce que les schemas font mieux que cette methode :**
un schema Pandera est du code versionne, re-executable a l'identique sur
n'importe quel futur export, en moins d'une seconde. Le resultat d'une IA
directe n'est pas automatiquement reproductible (nouvelle execution =
resultat potentiellement different) et doit etre relance manuellement.

```python
# Verification des 2 nouvelles regles trouvees par la methode 3
doublons_titre = df_web_products[df_web_products.duplicated(subset=["post_title"], keep=False)]
print(f"Fiches en doublon (meme titre, sku differents) : {len(doublons_titre)} fiches / {doublons_titre['post_title'].nunique()} titres")
display(doublons_titre[["sku", "post_title"]].sort_values("post_title"))

ventes_negatives = df_web_brut[df_web_brut["total_sales"] < 0]
print(f"\ntotal_sales negatif : {len(ventes_negatives)}")
display(ventes_negatives[["sku", "post_title", "total_sales"]])
```
